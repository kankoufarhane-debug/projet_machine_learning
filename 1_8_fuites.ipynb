## Chargement des données

```python
import pandas as pd

train = pd.read_csv("train.csv")        # sinistres avec la cible loss
test = pd.read_csv("a_estimer.csv")     # sinistres sans loss

# Variables d'entrée : toutes les colonnes sauf id et loss
variables = [c for c in train.columns if c not in ("id", "loss")]

# Les 14 continues : "cont1", ..., "cont14" (range(1, 15) s'arrête à 14)
cont = [f"cont{i}" for i in range(1, 15)]
```

## 1.8 Fuites

```python
# (a) id présents dans les deux fichiers : attendu 0
# set(...) = ensemble sans doublons ; & = intersection
print("id présents dans les deux fichiers :",
      len(set(train["id"]) & set(test["id"])))

# (b) id et loss sont-ils liés ?
# Spearman = corrélation de rang ; proche de 0 = pas de lien
print("corrélation id / loss (Spearman)   :",
      round(train["id"].corr(train["loss"], method="spearman"), 4))

# (c) lignes de a_estimer déjà présentes dans train
# hash_pandas_object : une empreinte par ligne, deux lignes identiques ont la même
h_train = pd.util.hash_pandas_object(train[variables], index=False)
h_test = pd.util.hash_pandas_object(test[variables], index=False)
print("lignes de a_estimer déjà vues dans train :", int(h_test.isin(h_train).sum()))

# (d) corrélation de chaque continue avec loss (les 5 plus fortes en valeur absolue)
corr = train[cont].corrwith(train["loss"], method="spearman")
ordre = corr.abs().sort_values(ascending=False).index
print()
print("Corrélations de rang avec loss (les 5 plus fortes) :")
print(corr.reindex(ordre).head(5).round(3))
```
