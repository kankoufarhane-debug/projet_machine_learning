## Chargement des données

```python
# pandas : bibliothèque qui manipule des tableaux de données
import pandas as pd

# read_csv lit un fichier CSV et crée un DataFrame (tableau en mémoire)
train = pd.read_csv("train.csv")        # sinistres avec la cible loss
test = pd.read_csv("a_estimer.csv")     # sinistres sans loss
```

## 1.1 Inventorier la table

```python
# Dictionnaire {nom: tableau} : une seule boucle pour les deux tables
tables = {"train.csv": train, "a_estimer.csv": test}

# Tout ce qui est décalé de 4 espaces sous le for est DANS la boucle
for nom, t in tables.items():
    print(f"=== {nom} ===")
    print("lignes, colonnes :", t.shape)                 # dimensions
    print("id uniques       :", t["id"].is_unique,       # la clé identifie-t-elle un sinistre ?
          f"(de {t['id'].min()} à {t['id'].max()})")
    print("cible présente   :", "loss" in t.columns)     # True dans train, False dans a_estimer
    print()
```
