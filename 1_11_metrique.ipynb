## Chargement des données

```python
import pandas as pd
import numpy as np

train = pd.read_csv("train.csv")        # sinistres avec la cible loss
y = train["loss"].to_numpy()            # la cible sous forme de liste de nombres
```

## 1.11 La métrique

### A. Le meilleur montant unique dépend de la métrique

```python
# MAE : moyenne des écarts en valeur absolue
def mae(y, c):
    return np.mean(np.abs(y - c))

# RMSE : racine de la moyenne des écarts AU CARRÉ (les gros écarts pèsent plus)
def rmse(y, c):
    return np.sqrt(np.mean((y - c) ** 2))

# Pour un montant unique c annoncé à tous les sinistres :
# la médiane minimise la MAE, la moyenne minimise la RMSE.
for nom, c in {"moyenne": y.mean(), "médiane": np.median(y)}.items():
    print(f"{nom:8s} c = {c:8.1f}   MAE = {mae(y, c):8.1f}   RMSE = {rmse(y, c):8.1f}")
```

### B. Le poids des gros sinistres dans chaque métrique

```python
c = np.median(y)
ecart = np.abs(y - c)                       # écart de chaque sinistre à la constante
gros = y >= np.quantile(y, 0.99)            # True pour les 1 % de sinistres les plus gros

# Part de l'erreur totale due à ces 1 % de sinistres
part_mae = ecart[gros].sum() / ecart.sum()                 # dans la MAE
part_rmse = (ecart[gros] ** 2).sum() / (ecart ** 2).sum()  # dans la RMSE (écarts au carré)

print(f"1 % des sinistres = {part_mae:.1%} de l'erreur MAE")
print(f"1 % des sinistres = {part_rmse:.1%} de l'erreur RMSE")
```
