## Chargement des données

```python
import pandas as pd

train = pd.read_csv("train.csv")        # sinistres avec la cible loss
test = pd.read_csv("a_estimer.csv")     # sinistres sans loss

# Variables d'entrée : toutes les colonnes sauf id et loss (130 au total)
variables = [c for c in train.columns if c not in ("id", "loss")]
print(len(variables), "variables d'entrée")
```

## 1.7 Doublons

```python
# (a) id en double : attendu 0
# .duplicated() = True pour chaque valeur qui répète une valeur déjà vue
print("id en double dans train     :", int(train["id"].duplicated().sum()))
print("id en double dans a_estimer :", int(test["id"].duplicated().sum()))

# (b) lignes identiques hors id (on compare les 130 variables d'entrée)
# keep=False marque TOUTES les lignes d'un groupe de doublons
masque = train.duplicated(subset=variables, keep=False)
print("lignes impliquées dans un groupe de doublons :", int(masque.sum()))
# Sans keep=False : seulement les copies "en trop"
print("copies en trop                               :",
      int(train.duplicated(subset=variables).sum()))

# (c) pour ces groupes, le montant est-il le même ?
if masque.any():    # True s'il existe au moins un doublon
    g = train[masque].groupby(variables)["loss"].agg(["count", "min", "max"])
    g["ecart"] = g["max"] - g["min"]       # écart de montant dans le groupe
    print("groupes où le montant diffère :", int((g["ecart"] > 0).sum()))
```
